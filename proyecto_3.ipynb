Para solucionarlo, vamos a crear una función llamada replace_wrong_values()



Define una función llamada replace_wrong_values() que reciba los siguientes parámetros:
df: el DataFrame a modificar
column: el nombre de la columna a trabajar
wrong_values: una lista con los valores incorrectos
correct_value: el valor correcto para reemplazar
Dentro de la función, usa un bucle for para iterar sobre cada valor incorrecto y aplicar .replace().


```python
def replace_wrong_values(df, column, wrong_values, correct_value):
    for wrong in wrong_values:
        df[column] = df[column].replace(wrong, correct_value)
```

Etapa 3.5

Ahora vamos a combinar dos criterios: día y ciudad.

Crea una función llamada number_tracks() que reciba dos parámetros:

day: un día de la semana (por ejemplo, 'Monday')
city: el nombre de una ciudad (por ejemplo, 'Springfield')
Dentro de la función:

Filtra el DataFrame por el día.
Luego, filtra por la ciudad.
Cuenta cuántas veces aparece 'user_id' en ese filtro.
Devuelve ese número como resultado.


```python
def number_tracks(day, city):
    # Almacena las filas donde el valor en 'day' es igual al parámetro day
    day_data = df[df['day'] == day]
    
    # Filtra las filas donde el valor en 'city' es igual al parámetro city
    city_data = day_data[day_data['city'] == city]
    
    # Extrae la columna 'user_id' y aplica el método count()
    track_count = city_data['user_id'].count()
    
    # Devuelve el número de valores de la columna 'user_id'
    return track_count
```

```python
# El número de canciones reproducidas en Springfield el lunes
springfield_lunes = number_tracks('Monday', 'Springfield')
print("Springfield el lunes:", springfield_lunes)
```

```python
# El número de canciones reproducidas en Shelbyville el lunes
shelbyville_lunes = number_tracks('Monday', 'Shelbyville')
print("Shelbyville el lunes:", shelbyville_lunes)
```

```python
# El número de canciones reproducidas en Springfield el viernes
springfield_viernes = number_tracks('Friday', 'Springfield')
print("Springfield el viernes:", springfield_viernes)
```

```python
# El número de canciones reproducidas en Shelbyville el viernes
shelbyville_viernes = number_tracks('Friday', 'Shelbyville')
print("Shelbyville el viernes:", shelbyville_viernes)
```
